Router accuracy baseline

```python
import json
from pathlib import Path
from astra_swarm.router import classify_alert
from astra_swarm.cassette import cassette

golden = json.loads(Path("/content/astra-swarm/evals/week2_routing_golden.json").read_text())

with cassette("week2_router_golden_baseline"):
    correct = 0
    for item in golden:
        r = classify_alert(item["alert"])
        ok = r.alert_class.value == item["expected_class"]
        correct += ok
        print(f"{'✓' if ok else '✗'} pred={r.alert_class.value:<14} exp={item['expected_class']:<14}")
    print(f"\nBaseline: {correct}/{len(golden)} = {correct/len(golden):.0%}")
```
